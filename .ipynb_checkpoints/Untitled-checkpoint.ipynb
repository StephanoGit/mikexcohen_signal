```python
import scipy.io as sio
import mne

import numpy as np
import matplotlib.pyplot as plt
```

```python
mat_data = sio.loadmat('uANTS_intro_MATLABfiles/sampleEEGdata.mat', simplify_cells=True)
eeg = mat_data['EEG']

erp = np.mean(eeg['data'], axis=2)

chan2plot = 'fcz'

labels = [loc['labels'].lower() for loc in eeg['chanlocs']]
chan_idx = labels.index(chan2plot.lower())

```

```python
plt.figure(figsize=(10, 4))

plt.plot(eeg['times'], erp[chan_idx, :], 'k', linewidth=2) # 'k' makes the line black

plt.title(f"ERP for channel {chan2plot.upper()}")
plt.xlabel("Time (ms)")
plt.ylabel("Amplitude")
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```
