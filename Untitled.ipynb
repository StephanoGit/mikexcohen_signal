```python
import scipy.io as sio
import numpy as np
import mne
import matplotlib.pyplot as plt

from mne.simulation import simulate_sparse_stc, simulate_evoked
```

```python

eeg = sio.loadmat('uANTS_intro_MATLABfiles/sampleEEGdata.mat', simplify_cells=True)
eeg = eeg['EEG']
erp = np.mean(eeg['data'], axis=2) * 1e-6

labels = [loc['labels'] for loc in eeg['chanlocs']]
info = mne.create_info(ch_names=labels, sfreq=eeg['srate'], ch_types='eeg')
info.set_montage('colin27_1020')  

evoked = mne.EvokedArray(erp, info, tmin=eeg['times'][0] / 1000)

evoked.plot_topomap(times=0.3, time_format="Topomap at 300 ms", size=3,show_names=True)
evoked.plot(picks=['PO3'], titles={"eeg": "ERP for FCz"})

plt.show()
```

```python
import scipy.io as sio
import numpy as np
import matplotlib.pyplot as plt
import mne

mat_data = sio.loadmat('NTSA_simEEG/emptyEEG.mat', simplify_cells=True)

lf = mat_data['lf']
eeg = mat_data['EEG']

diploc = 300 
```

```python
# Set up the figure with two subplots side-by-side
fig = plt.figure(figsize=(12, 6))

# --- Subplot 1: 3D Brain Dipole Locations (Matches plot3) ---
ax1 = fig.add_subplot(121, projection='3d')
grid_loc = lf['GridLoc']

# Plot all dipoles (blue edges, yellow face)
ax1.scatter(grid_loc[:, 0], grid_loc[:, 1], grid_loc[:, 2], 
            c='y', edgecolors='b', marker='o', alpha=0.5, label='All dipoles')

# Highlight the selected dipole (red edges, black face, larger size)
ax1.scatter(grid_loc[diploc, 0], grid_loc[diploc, 1], grid_loc[diploc, 2], 
            c='k', edgecolors='r', marker='s', s=100, label='Selected dipole')

ax1.set_title('Brain dipole locations')
ax1.set_box_aspect([1, 1, 1]) # Matches MATLAB's 'axis square'

# --- Subplot 2: Signal dipole projection (Matches topoplotIndie) ---
ax2 = fig.add_subplot(122)

# Extract the gain for this dipole. 
# MATLAB uses -lf.Gain(:,1,diploc). Python uses 0 for the first orientation index.
dipole_projection = -lf['Gain'][:, 0, diploc]

# Create the MNE info object required for plotting the topomap
labels = [loc['labels'] for loc in eeg['chanlocs']]
info = mne.create_info(ch_names=labels, sfreq=eeg['srate'], ch_types='eeg')
info.set_montage('colin27_1020') 

# Plot the topomap projection
# vlim=(-40, 40) matches MATLAB's set(gca,'clim',[-1 1]*40)
mne.viz.plot_topomap(dipole_projection, info, axes=ax2, vlim=(-40, 40), 
                     cmap='RdBu_r', contours=0, sphere='eeglab')

ax2.set_title('Signal dipole projection')

plt.tight_layout()
plt.show()
```
